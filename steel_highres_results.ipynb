## Training source

Archived source used for this run. The cell below was not executed in this notebook.

```python
#!/usr/bin/env python3
"""High-resolution binary steel-defect segmentation only.
Dependencies: torch torchvision segmentation-models-pytorch pandas numpy pillow scikit-learn matplotlib
"""
import argparse
import json
import random
from pathlib import Path
import numpy as np
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
import torch.nn.functional as F
import segmentation_models_pytorch as smp
from PIL import Image
from sklearn.model_selection import train_test_split
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from scipy import ndimage
from matplotlib.patches import Rectangle

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

def decode_rle(rle, height=256, width=1600):
    mask = np.zeros(height * width, dtype=np.uint8)
    if pd.isna(rle) or not str(rle).strip():
        return mask.reshape((height, width), order='F')
    values = np.asarray(str(rle).split(), dtype=int)
    starts = values[::2] - 1
    lengths = values[1::2]
    for start, length in zip(starts, lengths):
        mask[start:start + length] = 1
    return mask.reshape((height, width), order='F')

class SteelDataset(Dataset):

    def __init__(self, image_dir, df, resolution='high', augment=False):
        self.augment = augment
        self.color_jitter = transforms.ColorJitter(brightness=0.15, contrast=0.15)
        self.image_paths = sorted(Path(image_dir).glob('*.jpg'))
        self.annotations = {name: group for name, group in df.groupby('ImageId')}
        if resolution not in ('low', 'high'):
            raise ValueError("resolution must be 'low' or 'high'")
        self.size = (128, 800) if resolution == 'low' else (256, 1600)
        self.image_transform = transforms.Compose([transforms.Resize(self.size), transforms.ToTensor(), transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])])
        self.mask_transform = transforms.Resize(self.size, interpolation=transforms.InterpolationMode.NEAREST)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        path = self.image_paths[idx]
        with Image.open(path) as img:
            rgb = img.convert('RGB')
            if self.augment:
                rgb = self.color_jitter(rgb)
            image = self.image_transform(rgb)
        mask = np.zeros((256, 1600), dtype=np.uint8)
        if path.name in self.annotations:
            for _, row in self.annotations[path.name].iterrows():
                mask = np.maximum(mask, decode_rle(row['EncodedPixels']))
        mask = torch.from_numpy(mask.copy()).unsqueeze(0).float()
        mask = self.mask_transform(mask)
        if self.augment:
            if torch.rand(()) < 0.5:
                image, mask = image.flip(-1), mask.flip(-1)
            if torch.rand(()) < 0.5:
                image, mask = image.flip(-2), mask.flip(-2)
        return (image, mask)

def dice_loss(logits, targets, smooth=1.0):
    probs = torch.sigmoid(logits)
    intersection = (probs * targets).sum(dim=(1, 2, 3))
    denominator = probs.sum(dim=(1, 2, 3)) + targets.sum(dim=(1, 2, 3))
    dice = (2 * intersection + smooth) / (denominator + smooth)
    return 1 - dice.mean()

def segmentation_loss(logits, targets):
    bce = F.binary_cross_entropy_with_logits(logits, targets)
    dice = dice_loss(logits, targets)
    return bce + dice

def train_epoch(model, loader, optimizer):
    model.train()
    total_loss = 0.0
    for images, masks in loader:
        images = images.to(DEVICE)
        masks = masks.to(DEVICE)
        optimizer.zero_grad()
        logits = model(images)
        loss = segmentation_loss(logits, masks)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * images.size(0)
    return total_loss / len(loader.dataset)

@torch.no_grad()
def validate_epoch(model, loader):
    model.eval()
    total_loss = 0.0
    tp = fp = fn = 0
    for images, masks in loader:
        images = images.to(DEVICE)
        masks = masks.to(DEVICE)
        logits = model(images)
        loss = segmentation_loss(logits, masks)
        total_loss += loss.item() * images.size(0)
        predictions = torch.sigmoid(logits) >= 0.5
        targets = masks.bool()
        tp += (predictions & targets).sum().item()
        fp += (predictions & ~targets).sum().item()
        fn += (~predictions & targets).sum().item()
    precision = tp / (tp + fp + 1e-08)
    recall = tp / (tp + fn + 1e-08)
    dice = 2 * tp / (2 * tp + fp + fn + 1e-08)
    return {'loss': total_loss / len(loader.dataset), 'precision': precision, 'recall': recall, 'dice': dice}

@torch.no_grad()
def threshold_sweep(model, loader):
    model.eval()
    thresholds = np.round(np.arange(0.1, 0.951, 0.05), 2)
    counts = np.zeros((len(thresholds), 3), dtype=np.float64)
    for images, masks in loader:
        probabilities = torch.sigmoid(model(images.to(DEVICE)))
        targets = masks.to(DEVICE).bool()
        for i, threshold in enumerate(thresholds):
            predictions = probabilities >= float(threshold)
            counts[i] += [(predictions & targets).sum().item(),
                          (predictions & ~targets).sum().item(),
                          (~predictions & targets).sum().item()]
    rows = []
    for threshold, (tp, fp, fn) in zip(thresholds, counts):
        rows.append(dict(threshold=float(threshold),
                         precision=tp/(tp+fp+1e-8), recall=tp/(tp+fn+1e-8),
                         dice=2*tp/(2*tp+fp+fn+1e-8)))
    return pd.DataFrame(rows)


def save_training_plots(history, out):
    if not history:
        return
    frame = pd.DataFrame(history)
    fig, axes = plt.subplots(1, 3, figsize=(16, 4))
    axes[0].plot(frame.epoch, frame.train_loss, label="Training")
    axes[0].plot(frame.epoch, frame.loss, label="Validation")
    axes[0].set_title("Loss")
    for metric in ("precision", "recall", "dice"):
        axes[1].plot(frame.epoch, frame[metric], label=metric)
    axes[1].set_title("Validation metrics at threshold 0.5")
    axes[1].set_ylim(0, 1)
    axes[2].plot(frame.epoch, frame.lr, label="Learning rate")
    axes[2].set_yscale("log")
    axes[2].set_title("Learning rate")
    for ax in axes:
        ax.set_xlabel("Epoch")
        ax.legend()
    fig.tight_layout()
    fig.savefig(out / "training_curves.png", dpi=160)
    plt.close(fig)

@torch.no_grad()
def save_regions(model, dataset, indices, threshold, out, count):
    folder = out / "regions"
    folder.mkdir(exist_ok=True)
    model.eval()
    # Include both annotated and clean images where available.
    positives, negatives = [], []
    for i in indices:
        name = dataset.image_paths[i].name
        group = dataset.annotations.get(name)
        has_mask = group is not None and group.EncodedPixels.fillna("").astype(str).str.strip().ne("").any()
        (positives if has_mask else negatives).append(int(i))
    chosen = positives[:(count+1)//2] + negatives[:count//2]
    chosen += [int(i) for i in indices if int(i) not in chosen][:max(0, count-len(chosen))]
    mean = torch.tensor([0.485, 0.456, 0.406]).view(3,1,1)
    std = torch.tensor([0.229, 0.224, 0.225]).view(3,1,1)
    rows = []
    for i in chosen[:count]:
        image, mask = dataset[i]
        rgb = (image * std + mean).clamp(0,1).permute(1,2,0).numpy()
        truth = mask[0].numpy().astype(bool)
        probability = torch.sigmoid(model(image[None].to(DEVICE)))[0,0].cpu().numpy()
        prediction = probability >= threshold
        labels, total = ndimage.label(prediction)
        sizes = np.bincount(labels.ravel())
        fig, axes = plt.subplots(5, 1, figsize=(16, 12))
        axes[0].imshow(rgb)
        axes[0].set_title(dataset.image_paths[i].name)
        for ax, binary, title in ((axes[1], truth, "Ground-truth regions"),
                                  (axes[2], prediction, f"Predicted regions, threshold {threshold:.2f}")):
            ax.imshow(rgb)
            ax.imshow(np.ma.masked_where(~binary, binary), cmap="autumn", alpha=0.5, vmin=0, vmax=1)
            ax.set_title(title)
        for region, slices in enumerate(ndimage.find_objects(labels), start=1):
            if slices is None or sizes[region] < 25:
                continue
            ys, xs = slices
            axes[2].add_patch(Rectangle((xs.start, ys.start), xs.stop-xs.start, ys.stop-ys.start,
                                        fill=False, edgecolor="cyan", linewidth=1))
            rows.append(dict(image=dataset.image_paths[i].name, x=xs.start, y=ys.start,
                             width=xs.stop-xs.start, height=ys.stop-ys.start,
                             pixels=int(sizes[region]), threshold=threshold))
        errors = np.zeros((*truth.shape, 3), dtype=np.float32)
        errors[prediction & truth] = [0,1,0]
        errors[prediction & ~truth] = [1,0,0]
        errors[~prediction & truth] = [0,0.5,1]
        axes[3].imshow(errors)
        axes[3].set_title("Green: correct defect | Red: false positive | Blue: missed defect")
        axes[4].imshow(probability, cmap="magma", vmin=0, vmax=1)
        axes[4].set_title("Defect probability (0 to 1)")
        for ax in axes:
            ax.axis("off")
        fig.tight_layout()
        fig.savefig(folder / f"{dataset.image_paths[i].stem}_regions.png", dpi=140)
        plt.close(fig)
    pd.DataFrame(rows, columns=["image","x","y","width","height","pixels","threshold"]).to_csv(
        out / "predicted_regions.csv", index=False)
    print("Saved region visualizations:", folder, flush=True)

def run(args):
    if DEVICE != "cuda" and not args.allow_cpu:
        raise RuntimeError("GPU unavailable. Submit on a GPU node or use --allow-cpu.")
    random.seed(42)
    np.random.seed(42)
    torch.manual_seed(42)
    data = args.data_dir.expanduser().resolve()
    out = args.output_dir.expanduser().resolve()
    out.mkdir(parents=True, exist_ok=True)
    annotations = pd.read_csv(data / "train.csv")
    dataset = SteelDataset(data / "train_images", annotations, resolution="high")
    if len(dataset) < 2:
        raise ValueError("Expected JPG images inside train_images/")
    train_idx, val_idx = train_test_split(np.arange(len(dataset)), test_size=0.2,
                                        random_state=42, shuffle=True)
    train_dataset = SteelDataset(data / "train_images", annotations, resolution="high", augment=True)
    train_loader = DataLoader(Subset(train_dataset, train_idx), batch_size=args.batch_size,
                              shuffle=True, num_workers=args.workers,
                              pin_memory=DEVICE == "cuda")
    val_loader = DataLoader(Subset(dataset, val_idx), batch_size=args.batch_size,
                            shuffle=False, num_workers=args.workers,
                            pin_memory=DEVICE == "cuda")
    pd.DataFrame({"image": [dataset.image_paths[i].name for i in val_idx]}).to_csv(
        out / "validation_images.csv", index=False)
    model = smp.Unet(encoder_name="resnet34",
                     encoder_weights=None if args.checkpoint else "imagenet",
                     in_channels=3, classes=1, activation=None).to(DEVICE)
    if args.checkpoint:
        model.load_state_dict(torch.load(args.checkpoint.expanduser(),
                                         map_location=DEVICE, weights_only=True))
    optimizer = torch.optim.AdamW(model.parameters(), lr=args.lr, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="max", factor=0.5, patience=2, min_lr=1e-7)
    best = validate_epoch(model, val_loader)["dice"] if args.checkpoint else -1.0
    best_path = out / "highres_best.pth"
    if args.checkpoint:
        torch.save(model.state_dict(), best_path)
    history = []
    stale_epochs = 0
    print(f"Device: {DEVICE} | Training: {len(train_idx)} | Validation: {len(val_idx)}", flush=True)
    for epoch in range(args.epochs):
        loss = train_epoch(model, train_loader, optimizer)
        metrics = validate_epoch(model, val_loader)
        history.append(dict(epoch=epoch+1, train_loss=loss, lr=optimizer.param_groups[0]["lr"], **metrics))
        scheduler.step(metrics["dice"])
        pd.DataFrame(history).to_csv(out / "highres_history.csv", index=False)
        print(f"Epoch {epoch+1}/{args.epochs} | Train Loss: {loss:.4f} | "
              f"Val Loss: {metrics['loss']:.4f} | Precision: {metrics['precision']:.4f} | "
              f"Recall: {metrics['recall']:.4f} | Dice: {metrics['dice']:.4f}", flush=True)
        if metrics["dice"] > best:
            best = metrics["dice"]
            torch.save(model.state_dict(), best_path)
            stale_epochs = 0
        else:
            stale_epochs += 1
        if stale_epochs >= args.patience:
            print("Early stopping: no validation Dice improvement.", flush=True)
            break
    model.load_state_dict(torch.load(best_path, map_location=DEVICE, weights_only=True))
    results = threshold_sweep(model, val_loader)
    results.to_csv(out / "threshold_metrics.csv", index=False)
    # Default: choose balanced performance by maximum validation Dice.
    eligible = results[results.precision >= args.target_precision]
    if args.selection == "precision" and len(eligible):
        selected = eligible.sort_values(["recall", "dice"], ascending=False).iloc[0]
        reason = "Maximum validation recall meeting target precision"
    else:
        selected = results.loc[results.dice.idxmax()]
        reason = "Maximum validation Dice" if args.selection == "dice" else "Precision target not reached; using maximum validation Dice"
    summary = dict(selected.to_dict(), target_precision=args.target_precision,
                   selection=reason, best_checkpoint_dice_at_05=best)
    (out / "selected_threshold.json").write_text(json.dumps(summary, indent=2))
    print(results.to_string(index=False), flush=True)
    print(reason, selected.to_dict(), flush=True)
    plt.plot(results.threshold, results.precision, label="Precision")
    plt.plot(results.threshold, results.recall, label="Recall")
    plt.plot(results.threshold, results.dice, label="Dice")
    plt.xlabel("Threshold")
    plt.legend()
    plt.tight_layout()
    plt.savefig(out / "threshold_metrics.png", dpi=150)
    plt.close()
    save_training_plots(history, out)
    save_regions(model, dataset, val_idx, float(selected["threshold"]), out, args.examples)
    (out / "run_config.json").write_text(json.dumps({k: str(v) if isinstance(v, Path) else v for k, v in vars(args).items()}, indent=2))
    print("Completed successfully. Outputs:", out, flush=True)

if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--data-dir", type=Path, default=Path.home()/"finalproject")
    parser.add_argument("--output-dir", type=Path, default=Path("highres_outputs"))
    parser.add_argument("--epochs", type=int, default=20)
    parser.add_argument("--lr", type=float, default=3e-5)
    parser.add_argument("--batch-size", type=int, default=2)
    parser.add_argument("--workers", type=int, default=2)
    parser.add_argument("--target-precision", type=float, default=0.8)
    parser.add_argument("--checkpoint", type=Path, help="Existing ResNet34 U-Net state dictionary")
    parser.add_argument("--selection", choices=["dice", "precision"], default="dice")
    parser.add_argument("--patience", type=int, default=6)
    parser.add_argument("--examples", type=int, default=8)
    parser.add_argument("--allow-cpu", action="store_true")
    args = parser.parse_args()
    if args.epochs < 0 or (args.epochs == 0 and not args.checkpoint):
        parser.error("Use positive epochs, or --epochs 0 with --checkpoint for threshold tuning only")
    if args.patience < 1 or args.examples < 1 or args.batch_size < 1 or args.workers < 0 or not 0 < args.target_precision <= 1:
        parser.error("Invalid batch size, workers, or target precision")
    run(args)
```
